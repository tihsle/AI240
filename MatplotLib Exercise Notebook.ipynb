## Exercise 3 - Draw Barchart




```python
fruits = ["Apples", "Bananas", "Cherries", "Oranges"]
sales = [23, 45, 12, 34]

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(fruits, sales, color="skyblue")

ax.set_title("Fruit Sales Today")
ax.set_xlabel("Fruit")
ax.set_ylabel("Number Sold")

plt.show()
```

## Exercise 4 -  Draw Histogram


```python
scores = [
    55, 62, 67, 71, 73, 74, 75, 78, 81, 82, 
    83, 84, 85, 86, 87, 88, 89, 89, 90, 91, 
    92, 93, 94, 95, 95, 96, 97, 98, 99, 100
]

fig, ax = plt.subplots(figsize=(6, 4))

ax.hist(scores, bins=5, color="lightcoral", edgecolor="black")

ax.set_title("Distribution of Student Exam Scores")
ax.set_xlabel("Score Ranges")
ax.set_ylabel("Number of Students (Count)")

plt.show()
```

## Exercise 5 -  Draw Piechart



```python
fruits = ["Apples", "Bananas", "Cherries", "Oranges", "Grapes"]
proportions = [35, 25, 20, 15, 5]
colors = ["#ff4d4d", "#ffeb3b", "#ffa726", "#9c27b0", "#e91e63"]

explode = (0, 0, 0, 0, 0.1)

fig, ax = plt.subplots(figsize=(6, 6))

ax.pie(
    proportions,
    explode=explode,
    labels=fruits,
    colors=colors,
    autopct="%1.1f%%",
    shadow=True,
    startangle=140,
)

ax.set_title("Fruit Basket Composition")
plt.show()
```
